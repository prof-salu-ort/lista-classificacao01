```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report
```

```python
#1. Carregando os dados
dados_cancer = load_breast_cancer()
X_cancer = pd.DataFrame(dados_cancer.data, columns=dados_cancer.feature_names)
y_cancer = dados_cancer.target

X_cancer.head()
```

```python
#2. Divisão em treino (75%) e teste (25%)
X_train, X_test, y_train, y_test = train_test_split(X_cancer, y_cancer, test_size=0.25, random_state=42, stratify=y_cancer)
```

```python
#3. Treinamento de arvore com profundidade = 4
modelo_cancer = DecisionTreeClassifier(max_depth=5, random_state=42)
modelo_cancer.fit(X_train, y_train)
```

```python
#4. Previsões
y_pred = modelo_cancer.predict(X_test)
print(classification_report(y_test, y_pred, target_names=['Maligno (0)', 'Benigno (1)']))
```

```python
#5. Matriz
matriz_cancer = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=matriz_cancer, display_labels=['Maligno (0)', 'Benigno (1)'])

fig, ax = plt.subplots(figsize=(6,5))
disp.plot(cmap='Reds', ax=ax, values_format='d')
plt.title('Matriz de confusão: Diagnostico', fontsize=13, fontweight='bold')
plt.grid(False)
plt.show()
```
